```python
bruto = pd.read_excel(path, sheet_name="Sheet1", header=None)
```

```python
bruto
```

# Homens e mulheres por região com curso superior

```python
indicador_1 = bruto.iloc[4:9].copy()
```

```python
# Preenchendo os nomes das colunas
indicador_1.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
    "ctem_total",
    "ctem_homens",
    "ctem_mulheres",
    "educ_serv_saude_total",
    "educ_serv_saude_homens",
    "educ_serv_saude_mulheres",
]
```

```python
indicador_1 = indicador_1.reset_index(drop=True)

# valores são número, não texto
for col in indicador_1.columns[1:]:
    indicador_1[col] = pd.to_numeric(indicador_1[col], errors="coerce")

indicador_1 = indicador_1[["regiao", "homens", "mulheres"]]
indicador_1
```

```python
indicador_1.describe()
```

```python
indicador_1.describe().round(2)
```
